The statement **“Probability Theory is an Extension of Logic”** is **only partly true**. A mathematically safer statement is:

> Probability theory extends **some algebraic features of propositional logic** by replacing binary truth values `{0,1}` with degrees of belief or measures in `[0,1]`.

But it is **not completely true** that probability theory is simply an extension of logic, especially not of **First-Order Logic**, because probability and logic have different semantics, different notions of truth, and different inference behavior.

Also, one correction to the counterpoint: probability theory itself is not mathematically “stochastic” in the sense of being nondeterministic. The **axioms and calculations of probability are deterministic**. What is stochastic is the phenomenon being modeled, or the uncertainty represented.


## 1. Classical logic uses truth values; probability uses measures

In classical propositional or first-order logic, a statement is assigned a truth value:

$$
v(A) \in \{0,1\}
$$

where:

$$
1 = \text{true}, \qquad 0 = \text{false}
$$

For example, for any proposition $A$,

$$
v(A \lor \neg A)=1
$$

and

$$
v(A \land \neg A)=0
$$

By contrast, probability theory assigns a number:

$$
P(A) \in [0,1]
$$

So probability allows:

$$
P(A)=0.5
$$

which is neither true nor false in the classical logical sense.

That alone shows probability is not simply classical logic with extra notation. It changes the semantic range from:

$$
\{0,1\}
$$

to:

$$
[0,1]
$$

Truth valuation and probability measure are different mathematical objects.


## 2. Probability includes logic as a special case, but that does not make it identical to logic

There is one important sense in which the original statement is partly correct.

A classical truth valuation can be seen as a special kind of probability measure where every proposition gets probability either $0$ or $1$.

For example:

$$
P(A) \in \{0,1\}
$$

Then:

$$
P(\neg A)=1-P(A)
$$

If $P(A)=1$, then:

$$
P(\neg A)=0
$$

If $P(A)=0$, then:

$$
P(\neg A)=1
$$

So binary logic can be embedded inside probability theory as the extreme case where every proposition has probability $0$ or $1$.

But probability theory also permits cases like:

$$
P(A)=0.3, \qquad P(A)=0.5, \qquad P(A)=0.999
$$

Those are not truth values in classical logic. Therefore, probability theory is broader in one respect, but it is not merely logic.

A useful analogy:

$$
\text{Boolean truth values } \{0,1\} \subset [0,1]
$$

but that does **not** mean all reasoning in $[0,1]$ is just Boolean logic.


## 3. Logical connectives are truth-functional; probabilities are not

In classical logic, the truth value of $A \lor B$ is completely determined by the truth values of $A$ and $B$.

For example, if:

$$
v(A)=1, \qquad v(B)=0
$$

then:

$$
v(A \lor B)=1
$$

So logical disjunction is truth-functional.

But in probability theory, knowing only $P(A)$ and $P(B)$ does **not** determine $P(A \lor B)$.

The probability rule is:

$$
P(A \lor B)=P(A)+P(B)-P(A \land B)
$$

So you also need to know the overlap:

$$
P(A \land B)
$$

### Counterexample

Let:

$$
P(A)=0.5, \qquad P(B)=0.5
$$

Case 1: suppose $A$ and $B$ are mutually exclusive.

Then:

$$
P(A \land B)=0
$$

so:

$$
P(A \lor B)=0.5+0.5-0=1
$$

Case 2: suppose $A$ and $B$ overlap with probability $0.25$.

Then:

$$
P(A \land B)=0.25
$$

so:

$$
P(A \lor B)=0.5+0.5-0.25=0.75
$$

In both cases:

$$
P(A)=0.5, \qquad P(B)=0.5
$$

but:

$$
P(A \lor B)=1
$$

in one case, while:

$$
P(A \lor B)=0.75
$$

in the other.

Therefore, there is no universal function $f$ such that:

$$
P(A \lor B)=f(P(A),P(B))
$$

for all events $A,B$.

This is a major mathematical difference from classical logic. In logic, the truth of $A \lor B$ is determined by the truth of $A$ and $B$. In probability, the probability of $A \lor B$ is not determined by $P(A)$ and $P(B)$ alone.

That means probability does not merely extend the truth tables of logic.


## 4. Probability 1 is not the same as logical truth

In classical logic, if a proposition is true, its negation is impossible.

But in probability theory, an event can have probability $1$ without being logically necessary.

Example: let $X$ be uniformly distributed on the interval $[0,1]$.

Consider the event:

$$
A = \{X \neq 1/2\}
$$

Then:

$$
P(A)=1
$$

because the probability of hitting exactly $1/2$ is:

$$
P(X=1/2)=0
$$

But it is not logically impossible that:

$$
X=1/2
$$

The value $1/2$ is still inside the sample space $[0,1]$. It is possible in the mathematical model, but it has probability zero.

So:

$$
P(A)=1
$$

does not mean:

$$
A \text{ is logically necessary}
$$

And:

$$
P(A)=0
$$

does not always mean:

$$
A \text{ is logically impossible}
$$

This is one of the strongest mathematical reasons why probability is not identical to logic.

In probability theory, “almost sure” does not mean “logically true.”


## 5. Logical equivalence implies equal probability, but equal probability does not imply logical equivalence

If two propositions are logically equivalent, then they must have the same probability.

If:

$$
A \leftrightarrow B
$$

is logically valid, then:

$$
P(A)=P(B)
$$

But the reverse is false.

For example, in a fair coin toss:

$$
P(\text{Heads})=0.5
$$

and

$$
P(\text{Tails})=0.5
$$

So:

$$
P(\text{Heads})=P(\text{Tails})
$$

But the propositions are not logically equivalent. In fact, they are mutually exclusive:

$$
\text{Heads} \land \text{Tails} = \bot
$$

Equal probability does not mean same logical content.

Therefore, probability loses distinctions that logic preserves.

That again shows probability is not simply an extension of logic in the strict mathematical sense.


## 6. First-Order Logic is model-theoretic; probability requires additional measure structure

First-Order Logic deals with structures, domains, predicates, functions, and quantifiers.

For example:

$$
\forall x \, P(x)
$$

is either true or false in a given structure.

Suppose the domain is:

$$
D=\{a,b\}
$$

The statement:

$$
\forall x\, P(x)
$$

is true exactly when:

$$
P(a) \text{ is true and } P(b) \text{ is true}
$$

Classical First-Order Logic does not assign a probability to that sentence. It only asks whether it is true in a structure.

To get probabilities, you need extra structure, such as a probability distribution over possible worlds, interpretations, models, or domain elements.

For example, you could assign:

$$
P(P(a))=0.7
$$

or:

$$
P(\forall x\,P(x))=0.2
$$

but First-Order Logic itself does not determine those probabilities.

So probability is not automatically an extension of First-Order Logic. You must add something external:

$$
\text{Logic} + \text{measure/probability distribution}
$$

That is a new mathematical structure, not merely ordinary logic extended.


## 7. Logical inference is monotonic; probabilistic inference is not

Classical logic has monotonic consequence.

That means if:

$$
\Gamma \models A
$$

then adding more premises cannot invalidate the conclusion:

$$
\Gamma \cup \Delta \models A
$$

Example:

$$
\text{All humans are mortal}, \quad \text{Socrates is human}
$$

therefore:

$$
\text{Socrates is mortal}
$$

Adding more true premises does not undo the logical consequence.

But probabilistic reasoning behaves differently.

You might have:

$$
P(\text{flies} \mid \text{bird})=0.95
$$

but:

$$
P(\text{flies} \mid \text{bird} \land \text{penguin})=0.01
$$

Adding more information changes the probability dramatically.

So probabilistic inference is often non-monotonic, while classical logical inference is monotonic.

This is another major mathematical difference.


## 8. A compact mathematical disproof of the strong claim

Suppose someone claims:

> Probability theory is simply an extension of classical logic.

Then the probabilistic value of a compound statement should behave like an extension of logical truth tables.

In particular, for disjunction, there should be some function:

$$
f:[0,1]^2 \to [0,1]
$$

such that:

$$
P(A \lor B)=f(P(A),P(B))
$$

for all events $A,B$.

But this is impossible.

Take two cases.

### Case 1

Let:

$$
P(A)=0.5, \qquad P(B)=0.5, \qquad P(A \land B)=0
$$

Then:

$$
P(A \lor B)=0.5+0.5-0=1
$$

So:

$$
f(0.5,0.5)=1
$$

### Case 2

Let:

$$
P(A)=0.5, \qquad P(B)=0.5, \qquad P(A \land B)=0.25
$$

Then:

$$
P(A \lor B)=0.5+0.5-0.25=0.75
$$

So:

$$
f(0.5,0.5)=0.75
$$

Contradiction.

The same input pair $(0.5,0.5)$ would require two different outputs:

$$
1 \neq 0.75
$$

Therefore, no such truth-functional probabilistic extension exists.

So probability theory is not simply classical logic generalized from $\{0,1\}$ to $[0,1]$.


## Final conclusion

The statement:

> “Probability Theory is an Extension of Logic”

is **partly true** only in a limited sense: classical Boolean logic can be embedded into probability theory by restricting probabilities to $0$ and $1$.

But it is **not completely true mathematically**, because:

$$
\text{truth} \neq \text{probability}
$$

$$
P(A)=1 \not\equiv A \text{ is logically necessary}
$$

$$
P(A)=0 \not\equiv A \text{ is logically impossible}
$$

$$
P(A)=P(B) \not\Rightarrow A \leftrightarrow B
$$

and

$$
P(A \lor B) \text{ is not determined by } P(A),P(B) \text{ alone}
$$

A more precise formulation would be:

> Probability theory generalizes certain Boolean algebraic aspects of propositional logic, but it is not a full mathematical extension of classical First-Order Logic. It requires additional measure-theoretic structure and has a different semantics of inference, uncertainty, and truth.
